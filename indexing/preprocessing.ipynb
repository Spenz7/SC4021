```python
import pandas as pd
```

```python
df = pd.read_csv('combined_kmeans_sentiment.csv')
url_pattern = r'^(http|https)://\S+|www\.\S+$'
df_clean = df[~df['text'].str.match(url_pattern, na=False)]
df_clean.to_csv('cleaned_data.csv', index=False, encoding='utf-8')
print(f"Removed {len(df) - len(df_clean)} rows containing only links.")
```
